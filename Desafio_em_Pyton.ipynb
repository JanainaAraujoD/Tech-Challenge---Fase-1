**Aplicação Prática no Dia a Dia da Empresa**

**Gatilho Logístico Automático:**
Se um pedido acumular 2 ou mais dias de atraso, o modelo classifica o cliente como **"Risco de Detrator"**

**Ação Preventiva no Atendimento: **
O sistema gera um alerta priorizado para a equipe de atendimento entrar em contato proativamente antes da entrega, alinhando expectativas e oferecendo um cupom de retratação.

**Mantendo a compra**:

A intervenção reduz a migração para a zona de detração, resgatando clientes para a faixa de neutros/promotores (onde a taxa de recompra salta de 0% para 38%–100%).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score


# 1. Carga dos Dados
df = pd.read_csv('desafio_nps_fase_1.csv')

# 2. Definição da Target (1 = Detrator em Risco [NPS <= 6], 0 = Neutro/Promotor [NPS >= 7])
df['is_detractor'] = (df['nps_score'] <= 6).astype(int)

# 3. Seleção das informações para a pesquisa (apenas variáveis disponíveis antes do envio do NPS)
features = [
    'customer_age', 'customer_tenure_months', 'order_value', 'items_quantity',
    'discount_value', 'payment_installments', 'delivery_time_days',
    'delivery_delay_days', 'freight_value', 'delivery_attempts',
    'customer_service_contacts', 'resolution_time_days', 'complaints_count'
]

X = df[features]
y = df['is_detractor']

# 4. Divisão Treino e Teste (80% Treino, 20% Teste com estratificação)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 5. Treinamento do Modelo (Random Forest Classifier)
model = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
model.fit(X_train, y_train)

# 6. Avaliação dos Resultados
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("--- RELATÓRIO DE CLASSIFICAÇÃO ---")
print("")
print(classification_report(y_test, y_pred))
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_proba):.4f}")
```
